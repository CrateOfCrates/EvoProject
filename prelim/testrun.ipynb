```python
import torch
from evo2 import Evo2

# Verify CUDA visibility inside WSL
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# Load the model directly onto GPU
evo2_model = Evo2('evo2_7b')

# Generate DNA sequence on GPU
out = evo2_model.generate(
    prompt_seqs=["CCTCCCCCTC"], 
    n_tokens=20, 
    temperature=1.0
)
print(out.sequences[0])
```
